# Filtro tienda 44 — desde train_2016_2017_union.csv

Notebook independiente: solo carga `train_2016_2017_union.csv` (la base ya unida con stores, transactions, items, holidays_events y oil) y se queda con las filas de la **tienda 44**. No depende de haber corrido antes ningún otro notebook.

Como el archivo pesa ~6 GB, lo leemos por partes (*chunks*) y de cada parte nos quedamos solo con las filas de la tienda 44 — así nunca cargamos el archivo completo en memoria.

```python
import pandas as pd
import os

BASE_DIR = "C:/Tesis"
```

## Filtrado

```python
FILE_PATH = os.path.join(BASE_DIR, "train_2016_2017_union.csv")
CHUNK_SIZE = 1_000_000
STORE_NBR = 44

tamano_gb = os.path.getsize(FILE_PATH) / (1024**3)
print(f"Tamaño en disco: {tamano_gb:.2f} GB")
```

```python
partes = []

for i, chunk in enumerate(pd.read_csv(FILE_PATH, chunksize=CHUNK_SIZE, parse_dates=["date"])):
    partes.append(chunk[chunk["store_nbr"] == STORE_NBR])

    if (i + 1) % 5 == 0:
        print(f"Procesadas {(i + 1) * CHUNK_SIZE:,} filas leídas del archivo...")

bbdd_filtrada = pd.concat(partes, ignore_index=True)
print(f"\nFilas de la tienda 44: {len(bbdd_filtrada):,}")
```
